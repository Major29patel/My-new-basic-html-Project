**Linear Regression **
It is a simply and widely used statistical method to meodel the relationship betwen a dependent variable and independent variable . The goal is to find a line that best fits the data points.
Y = mx + c

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

house_size = np.array([1000, 1500,  1200 , 1400 , 1600]).reshape(-1,1)
house_price = np.array([1000000, 1500000, 1200000, 1400000, 1600000])
model = LinearRegression()
model.fit(house_size , house_price)
predicted_price = model.predict(house_size)
plt.scatter(house_size, house_price , color = 'blue' , label = 'Actual price')
plt.plot(house_size, predicted_price, color = 'red' , label = 'Predicted Price')
plt.xlabel("House size")
plt.ylabel('House Price')
plt.title("Linear Regression : House Price based on Price ")
plt.legend()
plt.show()
```

**Logistic Regression**
Logistic regression is a type of a machine learning algorithms used for predicting outcomes that are categoriacl , means the outcome falls into one of a few different catrgories. It is commonly used when the categorical are binary , True or False , Yes or NO , 1 or 0.




input = study_Hours
outcome  = Pass or Fail

**Tpyes of Logistic Regression**
1. Binary Logistic Regression
2. Multinomial Logistic Regression
3. Ordinal Logistic Regression

In logistic regression metrics are used to evaluate how well the model is performing .


**Accuracy :**
Accuarc is the percentage of correct prediction made by the model out of all predictions

**Precision :**  
The percentage of true positive prediction out of all positive prediction made by the model.


**F1 Score :**
The harmonic mean of precision and recall giving a balance betrween the two.
Formula for harmonic mean (2*a*b)/(a+b).

**Confusion Matrix :**
It is like a scoreboard that tells you how many your model got it right and how many times got it wrong for both negative and positive cases.


```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
# Sample Data [Age , income , purchased[1,0]]
data = np.array([
    [25,30 , 0],
    [25 , 40 , 0 ],
    [47 , 50 ,1],
    [52, 80 , 1],
    [46 , 70 , 1],
    [43 , 23 , 0],
    [33 , 90 , 1],
    [35 , 100, 1]
])
x = data[:,:2]
y = data[:,2]
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size = 0.25, random_state = 42)
model = LogisticRegression()
model.fit(x_train, y_train)
prediction = model.predict(x_test)
accuracy = accuracy_score(y_test, prediction)
print(f"Accuracy : {accuracy *100:2f}%")
```
